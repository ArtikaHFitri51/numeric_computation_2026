```python
EPS = 1e-12  # batas toleransi untuk dianggap nol


def cetak_matriks(A, judul=""):
    """Menampilkan matriks augmented dengan rapi."""
    if judul:
        print(judul)
    for baris in A:
        print("  [ " + "  ".join(f"{v:9.4f}" for v in baris[:-1]) + "  |  " + f"{baris[-1]:9.4f}" + " ]")
    print()


def eliminasi_gauss(A, tampilkan_langkah=True):
    """
    Menyelesaikan SPL dengan eliminasi Gauss + pivoting parsial.
    A : matriks augmented n x (n+1), berupa list of list
    Return: list solusi x, atau None jika tidak ada solusi unik
    """
    n = len(A)
    A = [[float(v) for v in baris] for baris in A]  # salin supaya input asli tidak berubah

    if tampilkan_langkah:
        cetak_matriks(A, "Matriks awal [A | b]:")

    # ---------- Tahap 1: Eliminasi maju ----------
    for k in range(n - 1):
        # pivoting parsial: cari baris p >= k dengan |A[p][k]| terbesar
        p = max(range(k, n), key=lambda i: abs(A[i][k]))

        if abs(A[p][k]) < EPS:
            print("Tidak ada solusi unik (pivot bernilai 0).")
            return None

        if p != k:
            A[k], A[p] = A[p], A[k]
            if tampilkan_langkah:
                cetak_matriks(A, f"Tukar baris {k+1} dengan baris {p+1}:")

        # eliminasi elemen di bawah pivot
        for i in range(k + 1, n):
            m = A[i][k] / A[k][k]
            for j in range(k, n + 1):
                A[i][j] -= m * A[k][j]

        if tampilkan_langkah:
            cetak_matriks(A, f"Setelah eliminasi kolom {k+1}:")

    if abs(A[n - 1][n - 1]) < EPS:
        print("Tidak ada solusi unik (a[n][n] bernilai 0).")
        return None

    # ---------- Tahap 2: Substitusi mundur ----------
    x = [0.0] * n
    x[n - 1] = A[n - 1][n] / A[n - 1][n - 1]
    for i in range(n - 2, -1, -1):
        jumlah = sum(A[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (A[i][n] - jumlah) / A[i][i]

    return x


def tampilkan_solusi(x):
    if x is None:
        return
    print("Solusi SPL:")
    for i, nilai in enumerate(x, start=1):
        print(f"  x{i} = {nilai:.6f}")


# ================== INPUT MANUAL ==================
def baca_bilangan(prompt):
    """Meminta input bilangan; mendukung pecahan seperti 1/2. Ulangi jika salah."""
    while True:
        teks = input(prompt).strip().replace(",", ".")
        try:
            if "/" in teks:
                atas, bawah = teks.split("/")
                return float(atas) / float(bawah)
            return float(teks)
        except (ValueError, ZeroDivisionError):
            print("  Input tidak valid, coba lagi.")


def input_matriks():
    """Input matriks augmented secara manual, baris per baris."""
    while True:
        try:
            n = int(input("Jumlah persamaan/variabel (n): "))
            if n >= 2:
                break
            print("  n minimal 2.")
        except ValueError:
            print("  Masukkan bilangan bulat.")

    print(f"\nMasukkan koefisien tiap persamaan ({n} koefisien + 1 konstanta),")
    print("dipisah spasi. Contoh untuk 2x + y - z = 8  ->  2 1 -1 8\n")

    A = []
    for i in range(n):
        while True:
            teks = input(f"Persamaan {i+1}: ").strip().replace(",", ".").split()
            if len(teks) != n + 1:
                print(f"  Harus tepat {n+1} bilangan, kamu isi {len(teks)}. Coba lagi.")
                continue
            try:
                baris = []
                for t in teks:
                    if "/" in t:
                        a, b = t.split("/")
                        baris.append(float(a) / float(b))
                    else:
                        baris.append(float(t))
                A.append(baris)
                break
            except (ValueError, ZeroDivisionError):
                print("  Ada input yang tidak valid, coba lagi.")
    return A


def contoh_bawaan():
    # 2x + y - z = 8
    # -3x - y + 2z = -11
    # -2x + y + 2z = -3
    # Solusi yang benar: x = 2, y = 3, z = -1
    return [
        [ 2,  1, -1,   8],
        [-3, -1,  2, -11],
        [-2,  1,  2,  -3],
    ]


# ================== PROGRAM UTAMA ==================
def main():
    while True:
        print("=" * 45)
        print("  SPL - Eliminasi Gauss + Pivoting Parsial")
        print("=" * 45)
        print("1. Input matriks sendiri")
        print("2. Pakai contoh bawaan")
        print("0. Keluar")
        pilihan = input("Pilih: ").strip()

        if pilihan == "0":
            print("Selesai.")
            break
        elif pilihan == "1":
            A = input_matriks()
        elif pilihan == "2":
            A = contoh_bawaan()
        else:
            print("Pilihan tidak dikenal.\n")
            continue

        print()
        langkah = input("Tampilkan langkah-langkah? (y/n): ").strip().lower() != "n"
        print()
        hasil = eliminasi_gauss(A, tampilkan_langkah=langkah)
        tampilkan_solusi(hasil)
        print()


if __name__ == "__main__":
    main()
```

# **Penyelesaian Sistem Persamaan Linear (SPL) dengan Metode Eliminasi Gauss dan Pivoting Parsial**

**Tujuan**

Program ini dibuat untuk menyelesaikan Sistem Persamaan Linear (SPL) menggunakan metode Eliminasi Gauss dengan pivoting parsial.

**Dasar Teori**

Metode Eliminasi Gauss merupakan metode penyelesaian SPL dengan mengubah matriks augmented menjadi matriks segitiga atas, kemudian mencari nilai variabel menggunakan substitusi mundur.

Pivoting parsial dilakukan dengan memilih elemen bernilai mutlak terbesar pada kolom yang sedang diproses, kemudian menukar baris jika diperlukan. Tujuannya adalah untuk meningkatkan kestabilan perhitungan numerik.

Secara umum, metode ini terdiri dari:

1.   Membentuk matriks augmented.
2.   Melakukan eliminasi maju.
1.   Melakukan substitusi mundur.
2.   Menampilkan hasil penyelesaian SPL.





```python
EPS = 1e-12  # Batas toleransi untuk dianggap nol
```

Variabel EPS digunakan sebagai batas toleransi untuk menentukan apakah suatu elemen matriks dianggap bernilai nol. Penggunaan toleransi diperlukan karena perhitungan komputer menggunakan bilangan floating-point yang dapat menghasilkan galat pembulatan.

```python
def cetak_matriks(A, judul=""):
    """Menampilkan matriks augmented dengan rapi."""
    if judul:
        print(judul)

    for baris in A:
        print(
            "  [ "
            + "  ".join(f"{v:9.4f}" for v in baris[:-1])
            + "  |  "
            + f"{baris[-1]:9.4f}"
            + " ]"
        )
    print()
```

Fungsi cetak_matriks() digunakan untuk menampilkan matriks augmented dalam format yang rapi. Bagian kiri garis vertikal menunjukkan koefisien variabel, sedangkan bagian kanan menunjukkan konstanta pada setiap persamaan.

Parameter judul digunakan untuk menampilkan keterangan pada setiap tahapan perhitungan. Format angka empat desimal digunakan agar hasil matriks lebih mudah dibaca.

```python
def eliminasi_gauss(A, tampilkan_langkah=True):
    """
    Menyelesaikan SPL dengan eliminasi Gauss + pivoting parsial.
    A : matriks augmented n x (n+1)
    Return: list solusi x, atau None jika tidak ada solusi unik
    """
    n = len(A)

    # Salin matriks agar input asli tidak berubah
    A = [[float(v) for v in baris] for baris in A]

    if tampilkan_langkah:
        cetak_matriks(A, "Matriks awal [A | b]:")

    # Tahap 1: Eliminasi maju
    for k in range(n - 1):

        # Pivoting parsial
        p = max(range(k, n), key=lambda i: abs(A[i][k]))

        if abs(A[p][k]) < EPS:
            print("Tidak ada solusi unik (pivot bernilai 0).")
            return None

        # Pertukaran baris jika diperlukan
        if p != k:
            A[k], A[p] = A[p], A[k]

            if tampilkan_langkah:
                cetak_matriks(
                    A,
                    f"Tukar baris {k+1} dengan baris {p+1}:"
                )

        # Eliminasi elemen di bawah pivot
        for i in range(k + 1, n):
            m = A[i][k] / A[k][k]

            for j in range(k, n + 1):
                A[i][j] -= m * A[k][j]

        if tampilkan_langkah:
            cetak_matriks(
                A,
                f"Setelah eliminasi kolom {k+1}:"
            )

    # Pemeriksaan pivot terakhir
    if abs(A[n - 1][n - 1]) < EPS:
        print("Tidak ada solusi unik (pivot terakhir bernilai 0).")
        return None

    # Tahap 2: Substitusi mundur
    x = [0.0] * n

    x[n - 1] = A[n - 1][n] / A[n - 1][n - 1]

    for i in range(n - 2, -1, -1):
        jumlah = sum(
            A[i][j] * x[j]
            for j in range(i + 1, n)
        )

        x[i] = (A[i][n] - jumlah) / A[i][i]

    return x
```

Fungsi eliminasi_gauss() merupakan bagian utama dari program. Fungsi ini menerima matriks augmented dan menyelesaikan SPL melalui dua tahapan utama.

**1. Eliminasi maju**

Program mencari elemen dengan nilai mutlak terbesar pada kolom pivot melalui pivoting parsial. Jika posisi elemen tersebut berbeda dari baris pivot, kedua baris ditukar. Selanjutnya, elemen-elemen di bawah pivot dieliminasi hingga terbentuk matriks segitiga atas.

**2. Substitusi mundur**

Setelah eliminasi maju selesai, nilai variabel dihitung mulai dari baris terakhir menuju baris pertama.

Parameter tampilkan_langkah digunakan untuk menentukan apakah perubahan matriks pada setiap tahapan akan ditampilkan.

Fungsi mengembalikan daftar nilai variabel jika solusi unik dapat dihitung. Jika pivot yang diperlukan bernilai mendekati nol, fungsi mengembalikan None. Pemeriksaan ini mendeteksi kondisi ketika solusi unik tidak dapat diperoleh, tetapi tidak membedakan secara lengkap antara sistem tanpa solusi dan sistem dengan banyak solusi.

```python
def tampilkan_solusi(x):
    if x is None:
        return

    print("Solusi SPL:")

    for i, nilai in enumerate(x, start=1):
        print(f"  x{i} = {nilai:.6f}")
```

Fungsi tampilkan_solusi() digunakan untuk menampilkan nilai setiap variabel yang diperoleh dari proses eliminasi Gauss. Fungsi menggunakan perulangan enumerate() untuk memberikan nomor variabel secara berurutan, seperti 𝑥1, 𝑥2, dan 𝑥3.

Hasil solusi ditampilkan dengan enam angka di belakang koma. Jika solusi bernilai None, fungsi tidak menampilkan nilai variabel.

```python
def baca_bilangan(prompt):
    """Meminta input bilangan; mendukung pecahan seperti 1/2."""
    while True:
        teks = input(prompt).strip().replace(",", ".")

        try:
            if "/" in teks:
                atas, bawah = teks.split("/")
                return float(atas) / float(bawah)

            return float(teks)

        except (ValueError, ZeroDivisionError):
            print("  Input tidak valid, coba lagi.")
```

Fungsi baca_bilangan() digunakan untuk membaca input numerik dari pengguna. Fungsi mendukung bilangan desimal, tanda koma sebagai pemisah desimal, dan bilangan pecahan seperti 1/2.

Jika input tidak valid atau penyebut pecahan bernilai nol, program akan menampilkan pesan kesalahan dan meminta pengguna memasukkan nilai kembali.

```python
def input_matriks():
    """Input matriks augmented secara manual, baris per baris."""
    while True:
        try:
            n = int(input("Jumlah persamaan/variabel (n): "))

            if n >= 2:
                break

            print("  n minimal 2.")

        except ValueError:
            print("  Masukkan bilangan bulat.")

    print(
        f"\nMasukkan koefisien tiap persamaan "
        f"({n} koefisien + 1 konstanta),"
    )
    print(
        "Dipisah spasi. Contoh: "
        "2x + y - z = 8 -> 2 1 -1 8\n"
    )

    A = []

    for i in range(n):
        while True:
            teks = (
                input(f"Persamaan {i+1}: ")
                .strip()
                .replace(",", ".")
                .split()
            )

            if len(teks) != n + 1:
                print(
                    f"  Harus tepat {n+1} bilangan, "
                    f"kamu isi {len(teks)}. Coba lagi."
                )
                continue

            try:
                baris = []

                for t in teks:
                    if "/" in t:
                        a, b = t.split("/")
                        baris.append(float(a) / float(b))
                    else:
                        baris.append(float(t))

                A.append(baris)
                break

            except (ValueError, ZeroDivisionError):
                print("  Ada input yang tidak valid, coba lagi.")

    return A
```

Fungsi input_matriks() digunakan untuk menerima matriks augmented dari pengguna secara manual. Pengguna terlebih dahulu menentukan jumlah persamaan dan variabel, kemudian memasukkan koefisien serta konstanta pada setiap persamaan.

Jika terdapat 𝑛 variabel, setiap baris harus berisi tepat 𝑛+1 bilangan. Program memeriksa jumlah dan validitas bilangan yang dimasukkan sebelum menyimpan baris tersebut ke dalam matriks.

```python
def contoh_bawaan():
    # 2x + y - z = 8
    # -3x - y + 2z = -11
    # -2x + y + 2z = -3
    # Solusi: x = 2, y = 3, z = -1

    return [
        [2, 1, -1, 8],
        [-3, -1, 2, -11],
        [-2, 1, 2, -3],
    ]
```

Fungsi contoh_bawaan() menyediakan matriks augmented yang dapat digunakan untuk menguji program tanpa perlu memasukkan koefisien secara manual.

Sistem persamaan yang digunakan adalah:

2𝑥+𝑦−𝑧=8

−3𝑥−𝑦+2𝑧=−11

−2𝑥+𝑦+2𝑧=−3

Solusi yang diharapkan adalah

𝑥=2, 𝑦=3, dan 𝑧=−1.

Contoh ini digunakan untuk memeriksa apakah program dapat menghasilkan solusi yang sesuai.

```python
def main():
    while True:
        print("=" * 45)
        print("  SPL - Eliminasi Gauss + Pivoting Parsial")
        print("=" * 45)

        print("1. Input matriks sendiri")
        print("2. Pakai contoh bawaan")
        print("0. Keluar")

        pilihan = input("Pilih: ").strip()

        if pilihan == "0":
            print("Selesai.")
            break

        elif pilihan == "1":
            A = input_matriks()

        elif pilihan == "2":
            A = contoh_bawaan()

        else:
            print("Pilihan tidak dikenal.\n")
            continue

        print()

        langkah = (
            input("Tampilkan langkah-langkah? (y/n): ")
            .strip()
            .lower() != "n"
        )

        print()

        hasil = eliminasi_gauss(
            A,
            tampilkan_langkah=langkah
        )

        tampilkan_solusi(hasil)
        print()
```

Fungsi main() berfungsi sebagai pengendali utama program. Pengguna dapat memilih untuk memasukkan matriks sendiri, menggunakan contoh bawaan, atau keluar dari program.

Setelah matriks dipilih, pengguna menentukan apakah langkah-langkah eliminasi ingin ditampilkan. Program kemudian memanggil fungsi eliminasi_gauss() untuk menghitung solusi dan tampilkan_solusi() untuk menampilkan hasilnya.

```python
if __name__ == "__main__":
    main()
```

Bagian ini digunakan untuk menjalankan fungsi main() ketika kode dieksekusi secara langsung. Dengan pemisahan ini, fungsi-fungsi sebelumnya dapat didefinisikan terlebih dahulu sebelum program utama dijalankan.

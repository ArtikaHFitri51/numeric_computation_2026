```python
EPS = 1e-12  # batas toleransi untuk dianggap nol


def cetak_matriks(A, judul=""):
    """Menampilkan matriks augmented dengan rapi."""
    if judul:
        print(judul)
    for baris in A:
        print("  [ " + "  ".join(f"{v:9.4f}" for v in baris[:-1]) + "  |  " + f"{baris[-1]:9.4f}" + " ]")
    print()


def eliminasi_gauss(A, tampilkan_langkah=True):
    """
    Menyelesaikan SPL dengan eliminasi Gauss + pivoting parsial.
    A : matriks augmented n x (n+1), berupa list of list
    Return: list solusi x, atau None jika tidak ada solusi unik
    """
    n = len(A)
    A = [[float(v) for v in baris] for baris in A]  # salin supaya input asli tidak berubah

    if tampilkan_langkah:
        cetak_matriks(A, "Matriks awal [A | b]:")

    # ---------- Tahap 1: Eliminasi maju ----------
    for k in range(n - 1):
        # pivoting parsial: cari baris p >= k dengan |A[p][k]| terbesar
        p = max(range(k, n), key=lambda i: abs(A[i][k]))

        if abs(A[p][k]) < EPS:
            print("Tidak ada solusi unik (pivot bernilai 0).")
            return None

        if p != k:
            A[k], A[p] = A[p], A[k]
            if tampilkan_langkah:
                cetak_matriks(A, f"Tukar baris {k+1} dengan baris {p+1}:")

        # eliminasi elemen di bawah pivot
        for i in range(k + 1, n):
            m = A[i][k] / A[k][k]
            for j in range(k, n + 1):
                A[i][j] -= m * A[k][j]

        if tampilkan_langkah:
            cetak_matriks(A, f"Setelah eliminasi kolom {k+1}:")

    if abs(A[n - 1][n - 1]) < EPS:
        print("Tidak ada solusi unik (a[n][n] bernilai 0).")
        return None

    # ---------- Tahap 2: Substitusi mundur ----------
    x = [0.0] * n
    x[n - 1] = A[n - 1][n] / A[n - 1][n - 1]
    for i in range(n - 2, -1, -1):
        jumlah = sum(A[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (A[i][n] - jumlah) / A[i][i]

    return x


def tampilkan_solusi(x):
    if x is None:
        return
    print("Solusi SPL:")
    for i, nilai in enumerate(x, start=1):
        print(f"  x{i} = {nilai:.6f}")


# ================== INPUT MANUAL ==================
def baca_bilangan(prompt):
    """Meminta input bilangan; mendukung pecahan seperti 1/2. Ulangi jika salah."""
    while True:
        teks = input(prompt).strip().replace(",", ".")
        try:
            if "/" in teks:
                atas, bawah = teks.split("/")
                return float(atas) / float(bawah)
            return float(teks)
        except (ValueError, ZeroDivisionError):
            print("  Input tidak valid, coba lagi.")


def input_matriks():
    """Input matriks augmented secara manual, baris per baris."""
    while True:
        try:
            n = int(input("Jumlah persamaan/variabel (n): "))
            if n >= 2:
                break
            print("  n minimal 2.")
        except ValueError:
            print("  Masukkan bilangan bulat.")

    print(f"\nMasukkan koefisien tiap persamaan ({n} koefisien + 1 konstanta),")
    print("dipisah spasi. Contoh untuk 2x + y - z = 8  ->  2 1 -1 8\n")

    A = []
    for i in range(n):
        while True:
            teks = input(f"Persamaan {i+1}: ").strip().replace(",", ".").split()
            if len(teks) != n + 1:
                print(f"  Harus tepat {n+1} bilangan, kamu isi {len(teks)}. Coba lagi.")
                continue
            try:
                baris = []
                for t in teks:
                    if "/" in t:
                        a, b = t.split("/")
                        baris.append(float(a) / float(b))
                    else:
                        baris.append(float(t))
                A.append(baris)
                break
            except (ValueError, ZeroDivisionError):
                print("  Ada input yang tidak valid, coba lagi.")
    return A


def contoh_bawaan():
    # 2x + y - z = 8
    # -3x - y + 2z = -11
    # -2x + y + 2z = -3
    # Solusi yang benar: x = 2, y = 3, z = -1
    return [
        [ 2,  1, -1,   8],
        [-3, -1,  2, -11],
        [-2,  1,  2,  -3],
    ]


# ================== PROGRAM UTAMA ==================
def main():
    while True:
        print("=" * 45)
        print("  SPL - Eliminasi Gauss + Pivoting Parsial")
        print("=" * 45)
        print("1. Input matriks sendiri")
        print("2. Pakai contoh bawaan")
        print("0. Keluar")
        pilihan = input("Pilih: ").strip()

        if pilihan == "0":
            print("Selesai.")
            break
        elif pilihan == "1":
            A = input_matriks()
        elif pilihan == "2":
            A = contoh_bawaan()
        else:
            print("Pilihan tidak dikenal.\n")
            continue

        print()
        langkah = input("Tampilkan langkah-langkah? (y/n): ").strip().lower() != "n"
        print()
        hasil = eliminasi_gauss(A, tampilkan_langkah=langkah)
        tampilkan_solusi(hasil)
        print()


if __name__ == "__main__":
    main()
```

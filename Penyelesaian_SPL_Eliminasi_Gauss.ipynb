3. **MODEL MATEMATIKA**

Berdasarkan data transaksi pembelian pada toko komputer, dapat di bentuk 3 persamaan linear dengan 3 variabel.

Misalkan:


*  x = harga satu keyboard (ribu rupiah).
*  y = harga satu mouse (ribu rupiah).
*  z = harga satu flashdisk (ribu rupiah).

Berdasarkan transaksi pertama, diperoleh persamaan:

                      2x + y + 2z = 170

Berdasarkan transaksi kedua, diperoleh persamaan:

                      x + 2y + z = 120

Berdasarkan transaksi ketiga, diperoleh persamaan:

                      3x + y + z = 160

Dengan demikian, diperoleh sistem persamaan linear tiga variabel yang akan diselesaikan menggunakan metode eliminasi gauss.

4. **MATRIKS AUGMENTED**

Setelah memperoleh model matematika, langkah selanjutnya adalah mengubah sistem persamaan linear tiga variabel menjadi matriks augmented.

Matriks augmented dibentuk dengan mengambil koefisien dari variabel x, y, dan z, serta nilai konstanta pada setiap persamaan.

Berdasarkan persamaan sebelumnya, diperoleh matriks augmented:

                      |2  1  2 | 170|
                      |1  2  1 | 120|
                      |3  1  1 | 160|

Matriks tersebut selanjutnya digunakan dalam proses penyelesaian menggunakan metode eliminasi Gauss.


5. **ALGORITMA PENYELESAIAN**

Algoritma penyelesaian sistem persamaan linear tiga variabel menggunakan metode eliminasi Gauss adalah sebagai berikut:

*   Mulai.
*   Masukkan koefisien persamaan linear tiga variabel.
*   Bentuk matriks augmented dari sistem persamaan.
*   Tentukan elemen pivot pada setiap kolom.
*   Lakukan pertukaran baris apabila elemen pivot bernilai nol.
*   Lakukan operasi baris elementer untuk mengubah matriks menjadi bentuk segitiga atas.
*   Lakukan substitusi mundur untuk memperoleh nilai variabel z, y, dan x.
*   Tampilkan hasil penyelesaian.
*   Selesai.

6. **IMPLEMENTASI ALGORITMA MENGGUNAKAN PYTHON**

Program berikut dibuat untuk menyelesaikan sistem persamaan linear tiga variabel menggunakan metode eliminasi Gauss.

```python

# Program Penyelesaian SPLTV Metode Eliminasi Gauss

# 1. Input Persamaan
print("PENYELESAIAN SPLTV METODE ELIMINASI GAUSS")
print("Masukkan koefisien setiap persamaan:")
print("Format: x y z hasil")
print("Contoh: 2 1 2 170\n")

A = []

for i in range(3):
    baris = list(map(float, input(f"Persamaan ke-{i+1}: ").split()))

    if len(baris) != 4:
        raise ValueError("Masukkan tepat 4 angka!")

    A.append(baris)

n = 3

# 2. Fungsi untuk menampilkan matriks
def tampilkan_matriks(A):
    for baris in A:
        print("  ".join(f"{x:8.0f}" for x in baris))

# 3. Menampilkan Matriks Augmented Awal
print("\nMATRIKS AUGMENTED AWAL")
tampilkan_matriks(A)

# 4. Proses Eliminasi Gauss
print("\nPROSES ELIMINASI GAUSS")

for i in range(n):

    # Memilih pivot terbesar
    pivot = max(range(i, n), key=lambda r: abs(A[r][i]))

    if abs(A[pivot][i]) < 1e-10:
        raise ValueError("Sistem tidak memiliki solusi tunggal.")

    # Pertukaran baris jika diperlukan
    A[i], A[pivot] = A[pivot], A[i]

    # Eliminasi baris di bawah pivot
    for j in range(i + 1, n):
        faktor = A[j][i] / A[i][i]

        print(f"R{j+1} = R{j+1} - ({faktor:.2f} x R{i+1})")

        for k in range(i, n + 1):
            A[j][k] -= faktor * A[i][k]

        tampilkan_matriks(A)

# 5. Menampilkan Hasil Eliminasi
print("\nMATRIKS SETELAH ELIMINASI GAUSS")
tampilkan_matriks(A)

# 6. Proses Substitusi Mundur
print("\nPROSES SUBSTITUSI MUNDUR")

hasil = [0.0, 0.0, 0.0]

for i in range(n - 1, -1, -1):
    jumlah = 0

    for j in range(i + 1, n):
        jumlah += A[i][j] * hasil[j]

    hasil[i] = (A[i][n] - jumlah) / A[i][i]

    if i == 2:
        print(f"z = {hasil[i]:.2f}")
    elif i == 1:
        print(f"y = {hasil[i]:.2f}")
    else:
        print(f"x = {hasil[i]:.2f}")

# 7. Menampilkan Hasil Akhir
print("\nHASIL AKHIR")
print(f"Harga Keyboard (x) = {hasil[0]:.4f}")
print(f"Harga Mouse (y) = {hasil[1]:.4f}")
print(f"Harga Flashdisk (z) = {hasil[2]:.4f}")
```

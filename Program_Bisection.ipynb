**PROGRAM PENYELESAIAN PERSAMAAN NONLINEAR MENGGUNAKAN METODE BISECTION**

**TUJUAN:**

Program ini bertujuan untuk mencari akar persamaan nonlinear dengan menggunakan metode Bisection atau metode bagi dua.

Persamaan yang digunakan adalah:
                            
                                                        f(x) = x^3 - x - 2

Data awal:
*   Batas bawah (a) = 1
*   Batas atas (b) = 2
*   Toleransi = 0,0001

**Metode Penyelesaian:**

Metode Bisection dilakukan dengan membagi interval menjadi 2 bagian untuk mendapatkan perkiraan akar persamaan. Proses dilakukan secara berulang hingga memenuhi batas toleransi yang ditentukan.

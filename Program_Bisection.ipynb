**PROGRAM PENYELESAIAN PERSAMAAN NONLINEAR MENGGUNAKAN METODE BISECTION**

1. **TUJUAN:**

Program ini bertujuan untuk mencari akar persamaan nonlinear dengan menggunakan metode Bisection atau metode bagi dua.

Persamaan yang digunakan adalah -->
                            
                                                        f(x) = x³ - x - 2

Data awal:
*   Batas bawah (a) = 1
*   Batas atas (b) = 2
*   Toleransi = 0,0001

2. **METODE PENYELESAIAN:**

Metode Bisection dilakukan dengan membagi interval menjadi 2 bagian untuk mendapatkan perkiraan akar persamaan. Proses dilakukan secara berulang hingga memenuhi batas toleransi yang ditentukan.

3. ALGORITMA PENYELESAIAN

* Mulai.
* Tentukan fungsi f(x) = x³ - x - 2
* Tentukan batas bawah a = 1 dan batas atas b = 2.
* Tentukan toleransi sebesar 0,0001.
* Hitung nilai f(a) dan f(b).
* Periksa apakah f(a) x f(b).
* Jika tidak, interval tidak memenuhi syarat metode Bisection.
* Hitung titik tengah interval menggunakan rumus --> xr = a + b / 2
* Hitung nilai f(xr).
* Periksa tanda hasil fungsi untuk menentukan interval baru.
* Ulangi proses pembagian interval sampai setengah lebar interval kurang dari atau sama dengan toleransi.
* Tampilkan nilai pendekatan akar persamaan.
* Selesai.

Keterangan:
* a = batas bawah.
* b = batas atas.
* xr = titik tengah interval.
* Toleransi = batas ketelitian yang ditentukan

4. DATA & PARAMETER PENGUJIAN

Pada program metode Bisection ini, terdapat beberapa data awal yang digunakan sebagai dasar perhitungan, yaitu:

| Parameter | Nilai |
|---|---|
| Fungsi | x³ - x - 2 = 0 |
| Batas bawah (a) | 1 |
| Batas atas (b) | 2 |
| Toleransi | 0,0001 |
| Metode | Bisection |

Penjelasan:

Batas bawah dan batas atas digunakan untuk menentukan interval awal pencarian akar. Toleransi sebesar 0,0001 digunakan sebagai batas ketelitian untuk menghentikan proses iterasi.
